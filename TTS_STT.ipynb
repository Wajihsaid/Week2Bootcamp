```python
from google.colab import files
import speech_recognition as sr

# Upload your audio file
uploaded = files.upload()  # Select a .wav or .mp3 file

filename = list(uploaded.keys())[0]
print("Uploaded file:", filename)

# Convert audio to text
recognizer = sr.Recognizer()
with sr.AudioFile(filename) as source:
    audio = recognizer.record(source)

try:
    text = recognizer.recognize_google(audio)
    print("You said:", text)
except sr.UnknownValueError:
    print("Could not understand audio.")
except sr.RequestError as e:
    print("Google API request error:", e)
```

```python
import edge_tts
from IPython.display import Audio

async def tts():
    communicate = edge_tts.Communicate(
              text="Soldiers of Carthage, today we march not for glory, but for destiny. Rome will remember our strength.",
        voice="en-US-ChristopherNeural",
        rate="-15%",
        pitch="-12Hz"


        )
    await communicate.save("output2.mp3")

await tts()

Audio("output2.mp3")
```
